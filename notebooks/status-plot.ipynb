```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
time = []
rift_vel = []
force = []

with open("../status.txt") as f:
    for line in f:
        if "Final rift vel.: time" in line:
            time.append(line.split()[4])
            rift_vel.append(line.split()[7])
            force.append(line.split()[10])
```

```python
time = np.array(time, dtype=float)
rift_vel = np.array(rift_vel, dtype=float)
force = np.array(force, dtype=float)
```

```python
fig, axs = plt.subplots(2, 1, figsize=(8, 4))
axs[0].plot(time, rift_vel)
axs[0].set_title('Rift Velocity over Time')
axs[0].set_xlabel('Time')
axs[0].set_ylabel('Rift Velocity')
axs[0].grid()
axs[1].plot(time, force, color='orange')
axs[1].set_title('Force over Time')
axs[1].set_xlabel('Time')
axs[1].set_ylabel('Force')
axs[1].grid()
plt.tight_layout()
plt.show()
```
